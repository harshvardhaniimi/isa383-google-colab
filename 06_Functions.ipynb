## Part 1: What is a Function?

A function is a named sequence of statements that performs a computation. When you define a function, you specify the name and the sequence of statements. Later, you can "call" the function by name.

A function takes an argument and returns a result (called the return value). Functions help organize code, avoid repetition, and make programs more readable and maintainable.

Every function in Python returns a value. If no `return` statement is specified, the function returns `None` by default.

````python
# print is a function: it takes an argument and displays it
print("Hello, ISA 383!")
````

In the example above, `print` is the function name, `"Hello, ISA 383!"` is the argument passed to the function, and `print` does not return a value (it returns `None`).

````python
# type is a function: it takes an argument and returns its data type
type(32)
````

````python
type('a')
````

````python
type(False)
````

Here, `type` is the function name, `32` is the argument, and `int` is the return value of the function.

## Part 2: Built-in Functions

Python provides many built-in functions that you can use directly without needing to define them. Some commonly used ones include `print()`, `type()`, `len()`, `int()`, `float()`, `str()`, `input()`, `max()`, `min()`, `sum()`, `range()`, `abs()`, `round()`, and `sorted()`.

### `max()` and `min()`

These functions return the largest and smallest values from their arguments. They work with numbers, strings, and other comparable types.

````python
# max and min with numbers
print(f"Maximum of 3, 5, 2, 8, 1 is {max(3, 5, 2, 8, 1)}")
print(f"Minimum of 3, 5, 2, 8, 1 is {min(3, 5, 2, 8, 1)}")
````

````python
max([1, 5, 6, 7, 9, 11])
````

````python
# String comparison is based on ASCII values
# Uppercase letters have lower ASCII values than lowercase letters
"apple" < "Apple"
````

````python
# max and min also work with strings (compared character by character)
print(f"Maximum of 'apple', 'apricot' is {max('apple', 'apricot')}")
print(f"Minimum of 'apple', 'banana', 'cherry' is {min('apple', 'banana', 'cherry')}")
````

### `len()` and `ord()`

The `len()` function returns the length of a sequence (string, list, etc.). The `ord()` function returns the ASCII value of a single character. When applied to strings, `max()` and `min()` find the character with the highest and lowest ASCII value respectively.

````python
text = "Hello, ISA 383!"
print(f"Length of '{text}' is {len(text)}")

print(f"Max character in '{text}' is {max(text)} which has ASCII value {ord(max(text))}")
print(f"Min character in '{text}' is {min(text)} which has ASCII value {ord(min(text))}")
````

### `sum()`

The `sum()` function adds up all elements in an iterable (like a list). It requires an iterable as its argument, not individual numbers.

````python
sum([1, 2, 3, 4, 5])  # sum of a list of numbers
````

````python
try:
    sum(1, 2)
except TypeError as error:
    print("Expected TypeError:", error)
````

````python
max(1, 2)
````

## Part 6: Creating Your Own Functions

You can define your own functions using the `def` keyword. The general syntax is:

```python
def function_name(parameters):
    """
    Docstring: describes what the function does.
    """
    # Function body: code to execute
    return result  # Optional: return a value
```

Functions allow you to name a block of code and reuse it by calling that name with different arguments.

````python
# A simple function that greets a person
def greet(name):
    """Greets a person with the provided name."""
    return f"Hello, {name}!"

print(greet("ISA 383"))
print(greet("AUS"))
````

````python
# Functions can call other functions
def greet_2_people(name1, name2):
    """Greets two people with the provided names."""
    return greet(name1) + "\n" + greet(name2)

print(greet_2_people("Alice", "Brad"))
````

````python
import math

# Function to calculate area of a circle
def area_of_circle(radius):
    """Calculates the area of a circle given its radius."""
    return math.pi * (radius ** 2)

print(area_of_circle(5))
````

````python
# Function to calculate area of a rectangle
def area_of_rectangle(length, width):
    """Calculates the area of a rectangle given its length and width."""
    return length * width

print(area_of_rectangle(4, 6))
````

````python
# Function with a calculation
def exponent(a, b):
    """Calculate a raised to power b."""
    return a ** b

exponent(7, 3)
````

````python
# Function with conditional logic
def is_even(number):
    """Returns True if the number is even, False if odd."""
    return number % 2 == 0

print(f"Is 4 even? {is_even(4)}")
print(f"Is 7 even? {is_even(7)}")
````

````python
# Function with multiple conditions
def get_grade(score):
    """Returns the letter grade based on the score."""
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

print(f"Grade for 85: {get_grade(85)}")
print(f"Grade for 72: {get_grade(72)}")
````

### Building a Custom Random Function

You can combine modules with your own functions. Here we build a function that generates a random integer in a range using `random.random()` and `math.floor()`.

````python
import random
def random_integer_less_than(n):
    """Returns a random integer between 0 and n-1."""
    return math.floor(random.random() * n) #int(...)

random_integer_less_than(10)
````

````python
def random_between(a, b):
    """Generate a random integer between a and b (inclusive of a, exclusive of b)."""
    r = random.random()
    x = int(a + (b - a) * r)
    return x

random_between(5, 10)
````

## Part 7: Fruitful vs Void Functions

**Fruitful functions** return a value using the `return` statement. You can store their result in a variable and use it later.

**Void functions** do not return a value. They perform an action (like printing to the console) but do not send any value back to the caller. Functions without a `return` statement return `None` by default.

````python
# Fruitful function: get_grade returns a value we can store and use
result = get_grade(95)
print(f"Grade for 95: {result}")
````

````python
# Void function: this version of exponent prints but does not return
def exponent_void(a, b):
    """Calculate a raised to power b (void version)."""
    print(a ** b)

x = exponent_void(7, 3)
print(x)  # None, because the function did not return anything
````

````python
# print itself is a void function
result = print("Hello, ISA 383!")
print(f"Result of print function: {result}")  # None
````

## Part 8: Flow of Execution

When Python encounters a function definition, it does not execute the body immediately. The body only runs when the function is called. Execution then jumps into the function body, and after the function completes, returns to where it was called. This is useful to visualize when functions call other functions.

````python
print("Program starts here")

def first_function():
    print("Inside first_function")
    second_function()
    print("Back in first_function")

def second_function():
    print("Inside second_function")

print("About to call first_function")
first_function()
print("Program ends here")
````

The execution flow is: (1) program starts at the first statement, (2) function definitions are registered but not executed until called, (3) when `first_function()` is called, execution jumps to its body, (4) inside `first_function`, `second_function()` is called and execution jumps again, (5) after `second_function` completes, control returns to `first_function`, and (6) after `first_function` completes, control returns to the main program.

## Part 9: Why Use Functions?

Functions provide four key benefits:

**Modularity** breaks down complex problems into smaller, manageable pieces. **Reusability** lets you use the same function in different parts of the program or in other programs entirely. **Readability** makes code easier to understand by giving meaningful names to blocks of code. **Maintainability** makes updates easier since changes only need to be made in one place.

### Modularity Example

Calculate the cost of decorating a circular room given its radius and cost per square meter. By defining `area_of_circle` separately, the decoration cost function stays clean and focused.

````python
def cost_of_decoration(radius, cost_per_sqm):
    """Calculates the total cost of decorating a circular room."""
    area = area_of_circle(radius)  # reusing the function defined earlier
    total_cost = area * cost_per_sqm
    return total_cost

print(f"Cost of decorating a room with radius 5m at $10/sqm: ${cost_of_decoration(5, 10):.2f}")
````

### Reusability Example

A website needs to validate email format in multiple places: during user registration and during newsletter signup. By defining `validate_email` once, both processes can reuse it.

````python
def validate_email(email):
    """Simple email validation: checks for @ and . in the email."""
    return "@" in email and "." in email

def process_user_registration(email, username):
    """Registers a user after validating their email."""
    if validate_email(email):
        print(f"User {username} registered with email {email}")
        return True
    else:
        print(f"Invalid email: {email}")
        return False

def process_newsletter_signup(email):
    """Adds a user to the newsletter after validating their email."""
    if validate_email(email):
        print(f"Email {email} added to newsletter")
    else:
        print(f"Invalid email for newsletter: {email}")

# Same validate_email function used in multiple places
process_user_registration("user@example.com", "john_doe")
process_user_registration("user@exampl", "john_doe")
process_newsletter_signup("subscriber@email.com")
````
