## 2. Lists are Mutable

You access list elements using the bracket operator with an index, just like strings. However, unlike strings, **lists are mutable** - you can change, add, or remove elements after the list is created. Indices start at 0, and negative indices count backward from the end.

````python
# Accessing and modifying list elements
books = ['1984', '3 Body Problem', 'Fahrenheit 451']
print("Original collection:", books)
print("First book:", books[0])

# Update the second book - this would fail with a string!
books[1] = "The Anthropocene Reviewed"
print("Updated collection:", books)
````

````python
s = "Farenhite 451"

try:
    s[0] = "L"
except TypeError as error:
    print("Expected TypeError:", error)
````

````python
# Negative indices count from the end
books = ['The Great Gatsby', 'To Kill a Mockingbird', 'Pride and Prejudice']
print("Last book:", books[-1])
print("Second to last:", books[-2])
````

````python
# Reverse a list using slicing with step -1
books[::-1]
````

## 3. List Operations

Lists support three fundamental operations. The `+` operator concatenates two lists into a new list. The `*` operator repeats a list a specified number of times. The `in` operator checks whether an element exists in a list and returns `True` or `False`.

````python
# Concatenation with +
a = [1, 3, 5]
b = [2, 4, 6]
c = a + b
print("Concatenated:", c)
````

````python
# Repetition with *
d = [0] * 4
e = [1, 2, 3] * 3

print("d:", d)
print("e:", e)
````

````python
# Membership with in
library = ['Moby Dick', 'Jane Eyre', 'Wuthering Heights']
print('Jane Eyre' in library)
print('Macbeth' in library)
````

## 4. List Slices

The slice operator works identically to string slicing. The syntax is `list[start:end:step]` where `start` is inclusive, `end` is exclusive, and `step` is optional. Omitting `start` means from the beginning, omitting `end` means to the end, and `[:]` creates a copy of the entire list.

````python
# List slicing examples
t = ['a', 'b', 'c', 'd', 'e', 'f']

print("t[1:3]:", t[1:3])   # elements at index 1 and 2
print("t[:4]:", t[:4])     # first 4 elements
print("t[3:]:", t[3:])     # from index 3 to end
print("t[:]:", t[:])       # copy of entire list
````

````python
# Using step in slicing
t[0:5:2]  # every second element from index 0 to 4
````

````python
# Slice assignment - modify multiple elements at once
t = ['a', 'b', 'c', 'd', 'e', 'f']
t[1:3] = ['x', 'y']
print(t)
````

## 5. List Methods

Python provides several methods that operate on lists. Most list methods modify the list **in place** and return `None`. This is different from string methods which return new strings.

````python
# append() - adds a single element to the end
t = ['a', 'b', 'c']
t.append('d')
print(t)
````

````python
# Alternative: concatenation (creates new list)
t = ['a', 'b', 'c']
t = t + ['d']
print(t)
````

````python
# extend() - adds all elements from another list
t1 = ['a', 'b', 'c']
t2 = ['d', 'e']

t1.extend(t2)  # equivalent to t1 = t1 + t2 #### WRONG: t1 = t1.extend(t2)
print("t1:", t1)
print("t2:", t2)  # t2 is unchanged
````

````python
# sort() - arranges elements in ascending order
t = ['d', 'c', 'e', 'b', 'a']
t.sort()
print(t)

# sort() with reverse=True for descending order
num = [5, 3, 9, 12]
num.sort(reverse=True)
print(num)
````

````python
# Sorting strings - uppercase letters come before lowercase (ASCII order)
fruits = ['banana', 'apple', 'cherry', 'Strawberry']
fruits.sort()
print(fruits)
````

````python
l1 = ["100", "200", "Apple", "apple", "Banana", "cherry"]
l1.sort()
print(l1)
````

````python
l1 = ["-100", " 200", "Apple", "apple", "Banana", "cherry"]
l1.sort()
print(l1)
````

## 6. Deleting Elements

There are several ways to remove elements from a list, each suited for different situations.

````python
# pop() - removes and returns element at index
t = ['a', 'b', 'c']
x = t.pop(1)  # removes element at index 1
print("List:", t)
print("Popped element:", x)
````

````python
# pop() without argument removes last element
r = ['a', 'b', 'c']
s = r.pop()
print("List:", r)
print("Popped element:", s)
````

````python
# del - removes element at index, does not return it
t = ['a', 'b', 'c']
del t[1]
print(t)
````

````python
t = 5
del t
````

````python
try:
    print(t)
except NameError as error:
    print("Expected NameError:", error)
````

````python
# remove() - removes by value (first occurrence only)
books = ['Animal Farm', 'The Hobbit', 'The Catcher in the Rye', 'Animal Farm']
books.remove('Animal Farm')
print(books)  # second 'Animal Farm' is still there
````

````python
# del with slice - removes multiple elements
t = ['a', 'b', 'c', 'd', 'e', 'f']
del t[1:5]
print(t)
````

## 7. Built-in Functions for Lists

Python provides several built-in functions that work with lists. These are not methods (called with dot notation) but standalone functions that take a list as an argument.

````python
# Built-in functions for numeric lists
nums = [3, 41, 12, 9, 74, 15]

print(f"Length: {len(nums)}")
print(f"Maximum: {max(nums)}")
print(f"Minimum: {min(nums)}")
print(f"Sum: {sum(nums)}")
print(f"Average: {sum(nums)/len(nums)}")
````

````python
# len, max, min work on string lists too
# max/min use alphabetical order (ASCII)
words = ['Python', 'java', 'javascript', 'C++']

print(f"Number of words: {len(words)}")
print(f"First alphabetically: {min(words)}")
print(f"Last alphabetically: {max(words)}")
````

⚠️ **Note**: `sum()` only works with numeric lists. `max()` and `min()` cannot compare mixed types.

## 8. Lists and Strings

Lists and strings are both sequences, but they are different types. Python provides easy ways to convert between them.

````python
# list() converts string to list of characters
s = 'spam'
t = list(s)
print(t)
````

````python
# split() breaks string into list of words (default: whitespace)
sentence = 'Our language evolved as a way of gossiping.'
words = sentence.split()
print(words)
````

````python
# split() with custom delimiter
s = 'Earth-Mars-Jupiter-Neptune'
planets = s.split('-')
print(planets)

s2 = 'Sun;Proxima Centauri;Rigil Kentaurus'
stars = s2.split(';')
print(stars)
````

````python
s = "ahjhkjhjkkfmdsfdshgsakj"
s.split("k")
````

````python
# join() is the inverse of split()
# Note: join is a STRING method, called on the delimiter
words = ['One', 'of', "history's", 'few', 'iron', 'laws']
sentence = ' '.join(words)
print(sentence)
````

````python
# join() with different delimiters
chapters = ["Introduction", "Methods", "Results", "Discussion"]

table_of_contents = ' | '.join(chapters)
print(table_of_contents)

file_name = '-'.join(chapters)
print(file_name)
````

## 9. Identity, Equivalence, and Aliasing

Understanding how Python handles object identity is crucial when working with mutable objects like lists. Two objects can be **equivalent** (same content) without being **identical** (same object in memory). The `==` operator tests equivalence; the `is` operator tests identity.

````python
# With strings, Python may reuse objects (optimization)
a = 'banana'
b = 'banana'
print("a == b:", a == b)  # equivalent?
print("a is b:", a is b)  # identical?
````

````python
# With lists, each creation makes a new object
a = [1, 2, 3]
b = [1, 2, 3]
print("a == b:", a == b)  # equivalent: same content
print("a is b:", a is b)  # NOT identical: different objects
````

**Aliasing** occurs when two variables refer to the same object. With mutable objects, this can lead to unexpected behavior.

````python
# Aliasing: b = a does NOT create a copy
a = [1, 2, 3]
b = a  # b now refers to the SAME list as a
print("b is a:", b is a)
````

````python
# Changes through one alias affect the other
a = [1, 2, 3]
b = a
b[0] = 17  # modify through b
print("a:", a)  # a is also changed!
print("b:", b)
````

````python
# To avoid aliasing, create a copy
a = [1, 2, 3]
b = a.copy()  # or: b = a[:] or b = list(a)
print("b is a:", b is a)  # different objects now

b[0] = 17
print("a:", a)  # a is unchanged
print("b:", b)
````

⚠️ **Warning**: Aliasing with mutable objects is error-prone. When you need an independent copy of a list, use `.copy()`, slice `[:]`, or `list()`.

## Summary

Lists are mutable sequences that can hold elements of any type. You access elements using bracket notation with indices starting at 0. Negative indices count from the end. Lists support concatenation (`+`), repetition (`*`), and membership testing (`in`).

Key methods include `append()` to add single elements, `extend()` to add multiple elements, `sort()` to order elements, `pop()` to remove and return elements, and `remove()` to delete by value. Most methods modify the list in place and return `None`.

Built-in functions `len()`, `sum()`, `max()`, and `min()` work with lists. Use `split()` to convert strings to lists and `join()` for the reverse.

Because lists are mutable, aliasing (multiple variables referencing the same list) can cause unexpected behavior. Use `.copy()` when you need an independent copy.
