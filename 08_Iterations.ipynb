## 2. The `for` Loop

A `for` loop iterates over a sequence (list, string, range, etc.) and executes a block of code once for each item. It is called a "definite loop" because Python knows how many times it will run before it starts.

```python
for variable_name in collection:
    # do something with variable_name
```

Python handles the bookkeeping for you. You do not need to track indices or check for the end of the collection.

````python
x = 1

for i in range(5):
  x += 1
  print(x)
````

### 2.1 Looping Over a List

The most common use case. Python picks each item from the list in order, assigns it to the loop variable, and runs the indented block.

````python
friends = ['Alice', 'Tom', 'Sally']
for friend in friends:
    print('Happy New Year:', friend)
print('Done!')
````

**What happens step by step:**
1. Python sees `friends` has 3 items.
2. It assigns `'Alice'` to `friend` and runs the print statement.
3. It moves to `'Tom'`, reassigns `friend`, and repeats.
4. After all items are processed, execution moves to the line after the loop.

The variable name `friend` is arbitrary. You could call it `person`, `name`, or `x`. Choose something readable.

### 2.5 Combining Loops with Conditionals

Loops become powerful when paired with `if` statements. You can filter, count, categorize, or accumulate values from a list based on conditions.

## 3. Looping Over Dictionaries

Dictionaries store data as key-value pairs. There are three ways to loop through them, each giving you access to different parts of the data.

| Method | What you get |
|---|---|
| `dict.items()` | Both key and value |
| `dict.keys()` | Only the keys |
| `dict.values()` | Only the values |

````python
# Employee salary data
employee_salaries = {
    'Alice': 75000,
    'Bob': 82000,
    'Charlie': 68000,
    'Diana': 95000
}

# Loop through both keys and values using .items()
print("Employee Salary Report:")
total_payroll = 0

for name, salary in employee_salaries.items():
    total_payroll += salary
    print(f"{name}: ${salary:,}")

print(f"\nTotal payroll: ${total_payroll:,}")
print(f"Average salary: ${total_payroll / len(employee_salaries):,.0f}")
````

````python
# Loop through keys only
print("Employee names only:")
for name in employee_salaries.keys():
    print(name)
````

````python
# Loop through values only
print("Salary amounts only:")
for salary in employee_salaries.values():
    print(f"${salary:,}")
````

When you loop with `.items()`, you get both the key and the value in each iteration. This is the most common pattern for dictionaries in data analysis, where you typically need paired information.

## 4. The `while` Loop

A `while` loop keeps running as long as its condition is `True`. Use it when you do not know in advance how many iterations you need.

A `for` loop says "do this for each item." A `while` loop says "keep doing this until something changes."

Every `while` loop needs three components:
1. **Initialization** - set up variables before the loop.
2. **Condition** - the test that decides whether to continue.
3. **Update** - change something inside the loop so it eventually ends.

````python
# Countdown from 10 to 1
n = 10  # 1. Initialization
while n > 0:  # 2. Condition
    print(n)
    n = n - 1  # 3. Update (without this, infinite loop!)
print('Houston, we have a takeoff!')
````

**Flow of execution:**
1. Python checks: is `n > 0`? (10 > 0 is True, so enter the loop.)
2. It prints 10, then subtracts 1. Now `n` is 9.
3. It checks again: 9 > 0? True. Continues.
4. This repeats until `n` becomes 0.
5. Check: 0 > 0? False. The loop ends and execution continues after it.

````python
for i in range(10, 0, -1):
    print(i)
````

### 4.1 Practical Uses of `while` Loops

**Input Validation:** Keep asking the user for input until they provide something valid.

````python
# Keep asking until user enters a valid age
age = -1  # Initialize with invalid value
while age < 0 or age > 120:
    try:
        age = int(input("Enter your age (0-120): "))
        if age < 0 or age > 120:
            print("Please enter a realistic age.")
    except ValueError:
        print("Please enter a number.")
        age = -1  # Reset to continue loop

print(f"Thank you! Your age is {age}")
````

**Processing Until a Condition is Met:** Useful in financial modeling where you need to find "how long until a target is reached" rather than "what happens after X periods."

````python
# Simulating account growth until target is reached
account_balance = 1000
target_balance = 5000
monthly_return = 0.05  # 5% monthly return
months = 0

print(f"Starting balance: ${account_balance:,.2f}")
print(f"Target balance: ${target_balance:,.2f}")

while account_balance < target_balance:
    months += 1
    account_balance = account_balance * (1 + monthly_return)
    print(f"Month {months}: ${account_balance:,.2f}")

print(f"\nTarget reached in {months} months!")
````

## 5. Infinite Loops, `break`, and `continue`

An infinite loop runs forever because its condition never becomes `False`. This is usually a bug, but when controlled with `break`, it becomes a deliberate design pattern.

````python
# DO NOT RUN THIS CELL - it will loop forever
count = 0
while count < 10:
    count += 1
    if count == 5:
        continue;
    print(count)

    # Forgot to increment count!
````

Common causes of accidental infinite loops: forgetting to update the loop variable, updating it in the wrong direction, using the wrong comparison operator, or logic errors in complex conditions. If you accidentally create one, press the stop button in VS Code or Ctrl+C in the terminal.

### 5.1 `break` Statements

`break` immediately exits the loop, skipping all remaining iterations. It is often paired with `while True` to create loops that exit on a specific condition.

````python
# Menu-driven program using while True + break
while True:
    user_input = input("Enter a command (or 'quit' to exit): ")

    if user_input.lower() == 'quit':
        print("Goodbye!")
        break  # Exit the loop

    print(f"You entered: {user_input}")

print("Program ended.")
````

### 5.2 `continue` Statements

`continue` skips the rest of the current iteration and jumps to the next one. It is useful when you want to ignore certain values without stopping the loop entirely.

````python
# Print only odd numbers from 0 to 9
for i in range(10):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i)
````

### 5.3 Using `break` and `continue` with Real Data

````python
# Process transactions, skip invalid ones, stop at a sentinel value
transactions = [150.0, -20.0, 300.0, 0, 45.0, -999, 200.0]

total = 0
for t in transactions:
    if t == -999:
        print("Sentinel value found. Stopping.")
        break  # Stop processing entirely
    if t <= 0:
        print(f"Skipping invalid transaction: {t}")
        continue  # Skip this one, keep going
    total += t
    print(f"Processed: ${t:.2f}")

print(f"Total of valid transactions: ${total:.2f}")
````

## 6. Quick Review: Functions with Flexible Arguments

Before we move to exercises, let us review function argument types since some exercises will ask you to write functions.

A function can accept required arguments (must be provided), default arguments (have a fallback value), `*args` (collects extra positional arguments into a tuple), and `**kwargs` (collects extra keyword arguments into a dictionary).

Rule: once you give a parameter a default value, all parameters to its right must also have defaults.

````python
# Required arguments
def greet(name, msg):
    print(f"Hello {name}, {msg}")

greet("Aisha", "Good morning!")
````

````python
# Default arguments
def greet_default(name, msg="Have a nice day!"):
    print(f"Hello {name}, {msg}")

greet_default("ISA 383")  # Uses default message
greet_default("Sharjah", "Good morning!")  # Overrides default
````

````python
# *args: variable number of positional arguments
def flexible_greet(*names, greeting="Hello"):
    for name in names:
        print(f"{greeting}, {name}!")

flexible_greet("Alice", "Bob", "Charlie")
flexible_greet("Diana", greeting="Welcome")
````

````python
# **kwargs: variable number of keyword arguments
def print_info(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

print_info(name="Alice", role="Engineer", city="Sharjah")
````
