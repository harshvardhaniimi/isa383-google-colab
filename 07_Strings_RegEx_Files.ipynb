## 1. Strings: The Building Block

A string is a sequence of Unicode characters. Think of it as an ordered container you can index, slice, and iterate over. Strings are one of the most common data types you will encounter in business analytics, whether you are cleaning survey responses, parsing log files, or standardizing product names.

### 1.1 Indexing and Slicing

Strings use zero-based indexing: `s[0]` is the first character, `s[-1]` is the last. Slices follow the pattern `s[start:stop:step]` where `start` is inclusive and `stop` is exclusive. Omit any part to use defaults.

````python
s = "banana"
print("s:", s)
print("len(s):", len(s))
````

````python
print(f"{s[0]=}, {s[1]=}, {s[-1]=}")  # indexing
print(f"s[1:4]: {s[1:4]}")      # characters at index 1, 2, 3
print(f"s[1:4:2]: {s[1:4:2]}")  # every 2nd character in range 1 to 4
print(f"s[:3]: {s[:3]}")        # from start to index 3 (exclusive)
print(f"s[3:]: {s[3:]}")        # from index 3 to end
print(f"s[::2]: {s[::2]}")      # every 2nd character
print(f"s[::-1]: {s[::-1]}")    # reversed string
````

### 1.4 Common String Methods

Python strings come with a rich set of built-in methods. The most frequently used ones fall into a few categories: trimming whitespace, changing case, searching, splitting, and joining.

| Method | What it does |
|---|---|
| `strip()`, `lstrip()`, `rstrip()` | Remove whitespace (or given chars) from both ends, left, or right |
| `lower()`, `upper()`, `title()` | Change case |
| `startswith()`, `endswith()` | Test prefix or suffix |
| `find(sub)` | Returns index of first occurrence, or -1 if not found |
| `split(sep)` | Splits string into a list at each occurrence of `sep` |
| `join(iterable)` | Joins a list of strings with the string as separator |
| `replace(old, new)` | Replaces all occurrences of `old` with `new` |

**Stripping whitespace:**

````python
s = "  \tHello\n  "

print(repr(s.strip()))    # "Hello" - trims both sides
print(repr(s.lstrip()))   # "Hello\n  " - trims only left
print(repr(s.rstrip()))   # "  \tHello" - trims only right

# With specific characters: removes any of these chars from the ends
t = "xx--He--x--llo--xx"
print(t.strip("x-"))    # "Hello" - removes x or - from both ends
print(t.lstrip("x-"))   # "Hello--xx"
print(t.rstrip("x-"))   # "xx--Hello"
````

**Case methods and searching:**

````python
email = " Alice@WonderLand.com "
print(email.strip())   # removes surrounding spaces
print(email.lower())   # all lowercase
print(email.upper())   # all uppercase
print(email.title())   # title case
````

````python
# find() returns the index of a substring, or -1 if not found
email = email.strip()
at_pos = email.find('@')
print(f"Position of '@': {at_pos}")
print(f"Username: {email[:at_pos]}")
print(f"Domain: {email[at_pos+1:]}")
````

**Splitting and joining:**

````python
# split() breaks a string into a list
line = "  apple, banana ,  cherry ,date "
fruits = line.split(",")
print("After split:", fruits)
````

````python
# Strip whitespace from each element using a loop
fruits_clean = []
for f in fruits:
    fruits_clean.append(f.strip())
print("Cleaned:", fruits_clean)
````

````python
# join() does the reverse: combines a list into a string
joined = "; ".join(fruits_clean)
print("Joined:", joined)
````

Notice how `split()` and `join()` are inverses. `split()` goes from a single string to a list, and `join()` goes from a list back to a single string with a separator you choose.

## 2. Regular Expressions: When Simple Methods Are Not Enough

Python's `re` module lets you search, extract, and transform text using patterns. When you need to find all phone numbers in a document, validate email formats, or pull structured fields out of messy text, regex is the right tool.

### Core Ideas

Use raw strings for patterns: `r"\d+"` means one or more digits. The key functions are:
- `re.search(pattern, text)` finds the first match anywhere in the string.
- `re.findall(pattern, text)` returns all non-overlapping matches.
- `re.sub(pattern, replacement, text)` replaces matches.
- Capturing groups `(...)` extract parts. Named groups `(?P<name>...)` improve readability.
- Flags like `re.I` (ignore case) and `re.M` (multiline) modify behavior.

See https://regex101.com/ for interactive testing. Regex can be hard to write from scratch, so tools like regex101 or an LLM can help you build and verify patterns.

### 2.1 Useful Mini-Patterns

| Pattern | Meaning |
|---|---|
| `\d` | Any digit (0-9) |
| `\w` | Any word character (letter, digit, underscore) |
| `\s` | Any whitespace (space, tab, newline) |
| `\S` | Any non-whitespace character |
| `+` | One or more of the preceding |
| `*` | Zero or more of the preceding |
| `?` | Zero or one of the preceding |
| `.` | Any character except newline |
| `[A-Za-z]` | Character class (ranges) |
| `[^,]` | Negated class (anything except comma) |
| `^` | Start of string/line |
| `$` | End of string/line |
| `\b` | Word boundary |

### 2.2 `re.search()` and `re.findall()`

````python
import re

text = """
Order #A-1023: qty=4, price=$19.99
Order #B-77: qty=12, price=$3.50
"""

# re.search finds the FIRST match
m = re.search(r"#([A-Z]-\d+)", text)
print("First order code:", m.group(1))
````

````python
# re.findall returns ALL matches
all_quantities = re.findall(r"qty=(\d+)", text)
print("All quantities:", all_quantities)
````

### 2.3 Capturing Groups and Named Groups

Parentheses `()` create capturing groups that let you extract specific parts of a match. Named groups `(?P<name>...)` make patterns more readable.

````python
# Named groups to extract structured fields
pattern = re.compile(
    r"Order\s+#(?P<code>[A-Z]-\d+):\s+qty=(?P<qty>\d+),\s+price=\$(?P<price>\d+\.\d{2})"
)

for m in pattern.finditer(text):
    print(m.groupdict())
````

## 3. Files: Reading and Writing Text

File I/O is how your programs interact with data stored on disk. Use context managers (`with` statements) to handle files safely. They guarantee the file is closed even if an error occurs.

| Mode | Meaning |
|---|---|
| `"r"` | Read (default) |
| `"w"` | Write (overwrites existing content) |
| `"a"` | Append (adds to end) |
| `"x"` | Create exclusively (fails if file exists) |

We will use `pathlib.Path` for constructing file paths, which is the modern and cross-platform approach.

### 3.1 Reading Files

There are two common approaches: read all text at once, or read line by line. Line-by-line is preferred for large files since it does not load everything into memory.

### 3.2 Combining Files and Regex

A powerful pattern: read a file line by line, apply a regex to each line, and collect the results. This is the foundation of log parsing, data extraction, and text mining.

````python
# Extract all email addresses from lines that start with "From: "
email_pattern = re.compile(r"^From:\s+(\S+@\S+)", re.I)

emails = []
for line in lines:
    match = email_pattern.match(line)
    if match:
        emails.append(match.group(1))

print(f"Extracted {len(emails)} email addresses:")
for e in emails:
    print(f"  {e}")
````

How this regex works:
1. `^From:\s+` matches lines starting with "From:" followed by one or more whitespace characters.
2. `(\S+@\S+)` captures the email address - non-whitespace characters on both sides of the `@` symbol.
3. `re.I` makes the match case-insensitive.

````python
# Count how many emails are from umich.edu
umich_count = 0
for e in emails:
    if e.endswith("@umich.edu"):
        umich_count += 1
print(f"Emails from umich.edu: {umich_count}")
````
