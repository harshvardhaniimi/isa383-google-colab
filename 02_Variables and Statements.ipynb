## 3. Variable Naming Rules

**Must start with a letter or underscore:** `abc123` is valid, `123abc` is not.

**Can contain letters, numbers, underscores:** `abc_123`, `_abc` are valid.

**Cannot be a reserved keyword:** `if`, `for`, `while`, `True`, `False`, `None`, etc.

**Case sensitive:** `X` and `x` are different variables.

**Conventions:** Use `snake_case`, lowercase for variables, UPPERCASE for constants, meaningful names.

````python
x = 5
X = 10
print(x)  # 5
print(X)  # 10
````

````python
# `as` is a reserved word, so this text is invalid Python.
invalid_code = "as = 5"

try:
    compile(invalid_code, "<example>", "exec")
except SyntaxError as error:
    print("Expected SyntaxError:", error.msg)
````

## 6. Arithmetic Operators

| Op | Operation | Example |
|---|---|---|
| `+` | Addition | `5+3`→`8` |
| `-` | Subtraction | `5-3`→`2` |
| `*` | Multiplication | `5*3`→`15` |
| `/` | Division | `5/3`→approximately `1.67` |
| `//` | Floor Division | `5//3`→`1` |
| `%` | Modulus | `5%3`→`2` |
| `**` | Exponent | `5**3`→`125` |

````python
x, y = 5, 3
print(f"{x}+{y}={x+y}, {x}-{y}={x-y}, {x}*{y}={x*y}")
print(f"{x}/{y}={x/y}, {x}//{y}={x//y}, {x}%{y}={x%y}, {x}**{y}={x**y}")
````
