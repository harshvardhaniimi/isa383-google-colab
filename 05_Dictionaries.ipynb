## 1. What Is a Dictionary?

A **dictionary** stores data as pairs of a *key* and a *value*. Think of it like a contact list: each name (key) maps to a phone number (value). Keys must be unique, and you use them to retrieve values quickly.

````python
# Creating empty dictionaries
contacts = {}
print('Empty dictionary:', contacts)

# Using dict() gives the same result
contacts_alt = dict()
print('Also empty:', contacts_alt)
````

### Creating a Dictionary with Initial Data

Use curly braces `{}` and separate key-value pairs with commas. Keys and values can be different data types.

````python
# Dictionary of UAE emirates and their capitals
emirates_capitals = {
    'Abu Dhabi': 'Abu Dhabi City',
    'Dubai': 'Dubai City',
    'Sharjah': 'Sharjah City',
    'Ajman': 'Ajman City'
}
print(emirates_capitals)
````

## 2. Accessing Dictionary Values

Use the key inside square brackets `[]` to get the corresponding value. If the key is missing, Python raises a `KeyError`, so you must make sure the key exists.

````python
emirates_capitals
````

````python
print("Capital of Dubai:", emirates_capitals['Dubai'])
````

````python
emirates_capitals['RAK']
````

````python
try:
    print(f"The capital of RAK is {emirates_capitals['RAK']}")
except KeyError as error:
    print("Expected KeyError for missing key:", error)
````

### Safe Access with `get()`

`dict.get(key, default)` returns the value if the key exists; otherwise it returns the default (or `None` if omitted).

````python
print(emirates_capitals.get('Fujairah', "NA"))
````

````python
print(emirates_capitals.get('RAK'))
````

## 3. Adding, Updating, and Removing Entries

- Assign a value to a new key to add data.
- Assign to an existing key to update its value.
- Use `del` or `pop()` to remove entries.

````python
emirates_capitals
````

```
dict_name[key] = value
```

````python
# Add a new emirate
emirates_capitals['Fujairah'] = 'Fujairah City'

# Update an existing entry
emirates_capitals['Dubai'] = 'Dubai City'

# Remove with del
del emirates_capitals['Ajman']

print(emirates_capitals)
````

````python
# Remove with pop (returns the value)
removed = emirates_capitals.pop('Sharjah')
print('Removed:', removed)
print('Current dictionary:', emirates_capitals)
````

## 5. Nested Dictionaries

A nested dictionary is a dictionary where the values are themselves dictionaries. This allows you to create complex data structures that can represent more detailed information.

````python
# Nested dictionary for UAE hotels
dubai_hotels = {
    'Burj Al Arab': {
        'stars': 7,
        'rooms': 202,
        'seafront': True
    },
    'Atlantis': {
        'stars': 5,
        'rooms': 1544,
        'seafront': True
    },
    'Rove Downtown': {
        'stars': 3,
        'rooms': 356,
        'seafront': False
    }
}
````

````python
dubai_hotels["Atlantis"]
````

````python
type(dubai_hotels["Atlantis"])
````

````python
# Access specific info
print("Atlantis stars:", dubai_hotels['Atlantis']['stars'])
````

````python
print("Is Burj Al Arab seafront?", dubai_hotels['Burj Al Arab']['seafront'])
````

````python
dubai_hotels.get('Taj', {}).get('seafront', "Hotel doesn't exist")
````

````python
# Update nested values
dubai_hotels['Rove Downtown']['stars'] = 4
````

````python
dubai_hotels['Rove Downtown']
````

````python
dubai_hotels["Taj"] = ["List", "of", "things"]
print(dubai_hotels)
````
