## Part 1: Boolean Foundations and Comparison

### Boolean Values and Truthiness

Every conditional statement ultimately evaluates to one of two Boolean values: `True` or `False`. Python extends this concept through "truthiness" where any value can be evaluated in a Boolean context. Understanding what Python considers truthy or falsy is essential for writing clean conditional code.

**Falsy values** (evaluate to `False`):
- `None`
- Boolean `False`
- Numeric zeros: `0`, `0.0`, `0j`
- Empty sequences: `''`, `[]`, `()`, `{}`
- Empty sets: `set()`

**Truthy values**: Everything else evaluates to `True`.

````python
# Demonstrating falsy values
print("Falsy values:")
print(bool(False), bool(None), bool(0), bool(0.0), bool(''), bool([]))

# Demonstrating truthy values
print("\nTruthy values:")
print(bool(True), bool(42), bool(3.14), bool('Hello'), bool(' '))

# Practical application
inventory = []
if not inventory:  # Cleaner than: if len(inventory) == 0
    print("\nInventory is empty!")
````

### Comparison Operators

Comparison operators return Boolean values and are the building blocks of conditional logic.

**Operators:**
- `==` Equal to
- `!=` Not equal to
- `<` Less than
- `<=` Less than or equal to
- `>` Greater than
- `>=` Greater than or equal to

Python allows comparison chaining, which makes range checks more readable: `a < b < c` is equivalent to `a < b and b < c`.

````python
# Basic comparisons
price = 150
budget = 200

print(f"Can afford: {price <= budget}")
print(f"Exact match: {price == budget}")

# Comparison chaining for range checks
age = 25
print(f"\nWorking age (18-65): {18 <= age <= 65}")
````

````python
# String comparison (lexicographic order)
print(f"\n'apple' < 'banana': {'apple' < 'aanana'}")
print(f"'Apple' < 'apple': {'Apple' < 'apple'}")  # uppercase comes before lowercase
````

### Identity and Membership Operators

**Identity operators** (`is`, `is not`) check whether two variables reference the same object in memory, not just equal values.

**Membership operators** (`in`, `not in`) test whether a value exists in a sequence.

````python
# is vs ==
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print(f"list1 == list2: {list1 == list2}")  # Same values
print(f"list1 is list2: {list1 is list2}")  # Different objects
print(f"list1 is list3: {list1 is list3}")  # Same object
````

## Part 2: Conditional Statements and Logical Operators

### If-Elif-Else Structure

Conditional statements control program flow based on Boolean conditions. Python uses indentation (4 spaces) to define code blocks. The general structure:

```python
if condition1:
    # executed if condition1 is True
elif condition2:
    # executed if condition1 is False and condition2 is True
else:
    # executed if all conditions are False
```

Conditions are evaluated sequentially, and only the first true condition executes.

````python
# Simple grading system
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f"Score: {score}, Grade: {grade}")
````

### Logical Operators

Logical operators combine multiple conditions:
- `and`: True if both conditions are True
- `or`: True if at least one condition is True
- `not`: Inverts the Boolean value

**Short-circuit evaluation**: Python stops evaluating as soon as the result is determined. For `and`, if the first condition is False, the second isn't evaluated. For `or`, if the first condition is True, the second isn't evaluated.

````python
# Business rule: discount eligibility
purchase_amount = 1500
is_member = True
days_since_last = 45

# Multiple conditions with and
if purchase_amount > 1000 and is_member:
    print("Eligible for premium discount")

# Multiple conditions with or
if purchase_amount > 2000 or (is_member and days_since_last > 30):
    print("Eligible for welcome-back offer")

# Using not
if not is_member:
    print("Sign up to earn rewards!")
````

**Short-circuit demonstration:**

````python
# Safe division using short-circuit
denominator = 0
numerator = 10

# Second condition never evaluated when denominator is 0
if denominator != 0 and numerator / denominator > 1:
    print("Result is greater than 1")
else:
    print("Cannot divide or result not greater than 1")

# Using or for default values
user_name = ''
display_name = user_name or 'Guest'  # Returns 'Guest' since user_name is falsy
print(f"Welcome, {display_name}!")
````

### Nested Conditionals

Conditionals can be nested for complex logic. However, deep nesting reduces readability. Consider using logical operators or early returns instead.

````python
# Loan approval logic
credit_score = int(input("Credit score?"))
annual_income = 45000
has_collateral = True

# Nested approach (works but harder to read)
if credit_score >= 700:
    if annual_income >= 50000:
        print("Loan approved")
    else:
        if has_collateral:
            print("Loan approved with collateral")
        else:
            print("Loan denied - income too low")
else:
    print("Loan denied - credit score too low")

# Better approach using logical operators
if credit_score >= 700 and annual_income >= 50000:
    print("\nLoan approved (clean logic)")
elif credit_score >= 700 and has_collateral:
    print("\nLoan approved with collateral (clean logic)")
else:
    print("\nLoan denied (clean logic)")
````

## Part 3: Input Validation and Practical Applications

### Input Validation Principles

Robust programs validate all user input. Key validation steps:
1. Check data type (can it be converted?)
2. Check value range (is it within expected bounds?)
3. Check format (does it match expected pattern?)
4. Handle edge cases (empty input, extreme values)

Use guard clauses to check invalid conditions first and exit early.

````python
# Shipping cost calculator with validation
weight_input = input("Input weight:")
dest_input = "GCC"

# Type conversion with validation
try:
    weight = float(weight_input)
    print(f"{weight_input} is valid float")
except:
    print(f"Error: '{weight_input}' is not a valid number")
    weight = -1  # Mark as invalid
````

````python
weight = float(input("Weight in kgs? "))
dest_input = input("Destination? ")
# Guard clauses for validation
if weight <= 0:
    print(f"Invalid weight: {weight}. Must be positive.")
elif dest_input not in ['UAE', 'GCC', 'INTL']:
    print(f"Invalid destination: {dest_input}")
else:
    # Main logic only executes if validation passes
    if dest_input == "UAE":
        base = 10
        rate = 5
    elif dest_input == "GCC":
        base = 20
        rate = 10
    else:  # INTL
        base = 30
        rate = 15

    # Calculate charge
    charge = base if weight <= 5 else base + (weight - 5) * rate
    print(f"\nShipping to {dest_input} for {weight}kg: {charge} AED")
````

### Complete Business Application Example

Let's build a comprehensive commission calculator with multiple business rules.

````python
# Sales commission calculator
monthly_sales = 12000
years_employed = 3
is_top_performer = True

base_salary = 2000

# Validate inputs
if monthly_sales < 0:
    print("Error: Sales cannot be negative")
elif years_employed < 0:
    print("Error: Years employed cannot be negative")
else:
    # Calculate commission based on tiers
    if monthly_sales <= 5000:
        commission = 0
    elif monthly_sales <= 10000:
        # 5% on amount over 5000
        commission = (monthly_sales - 5000) * 0.05
    else:
        # 5% on first tier (5000-10000) + 10% on amount over 10000
        commission = 5000 * 0.05 + (monthly_sales - 10000) * 0.10

    # Seniority bonus (2% of sales if employed >= 3 years)
    seniority_bonus = monthly_sales * 0.02 if years_employed >= 3 else 0

    # Performance bonus (flat 500 for top performers)
    performance_bonus = 500 if is_top_performer else 0

    total_compensation = base_salary + commission + seniority_bonus + performance_bonus

    print(f"Sales: {monthly_sales} AED")
    print(f"Base Salary: {base_salary} AED")
    print(f"Commission: {commission} AED")
    print(f"Seniority Bonus: {seniority_bonus} AED")
    print(f"Performance Bonus: {performance_bonus} AED")
    print(f"Total Compensation: {total_compensation} AED")
````

## Common Pitfalls to Avoid

**1. Indentation Errors**: Python uses indentation to define blocks. Mix tabs and spaces causes errors. Use 4 spaces consistently.

**2. Assignment vs Comparison**: Using `=` instead of `==` in conditions is a syntax error in Python (unlike some languages).

**3. Identity vs Equality**: Use `is` for `None` checks, `==` for value comparisons.

**4. Floating Point Comparisons**: Avoid exact `==` checks for values produced by floating-point calculations when rounding error matters. Compare with a tolerance instead.

**5. Overly Complex Conditions**: Break complex logic into intermediate variables for clarity.

````python
# Pitfall: Floating point comparison
# Bad
value = 0.1 + 0.1 + 0.1
print(f"0.1 + 0.1 + 0.1 == 0.3: {value == 0.3}")  # False!

# Good
tolerance = 1e-9
print(f"Within tolerance: {abs(value - 0.3) < tolerance}")

# Pitfall: Complex condition (hard to read)
# Bad
if score >= 90 and attendance >= 0.8 and assignments_complete >= 10 and participation > 5:
    pass

# Good
has_good_score = score >= 90
has_good_attendance = attendance >= 0.8
completed_requirements = assignments_complete >= 10 and participation > 5

if has_good_score and has_good_attendance and completed_requirements:
    print("\nEligible for honors")
````
